# CSE-481 BIST 100 Agentic Analytics Harness - Classroom Demo

This notebook is a lightweight walkthrough for the final classroom demonstration. It does not contain API keys, raw cache files, or investment advice. Python modules calculate; the LLM explains committed evidence; the harness controls state, quality gates, human review, and replay.

## What This Project Demonstrates

- Fixed 30-stock BIST study universe and sector map.
- Four required research scenarios: sector catch-up, weekday/multi-day patterns, technical reversals, and quarterly fundamentals.
- Point-in-time validation for fundamentals, macro context, RSS news context, and historical decisions.
- Backtesting with costs, slippage, benchmark comparison, risk metrics, unseen split, and regime split.
- MCP-like deterministic tool surface, stateful harness, evidence bundle, quality gate, human review, and replayable decision logs.
- Optional live Gemini LLM harness explanation and a static dashboard UI.

## Known Limitations To State Clearly

- Fintables is not used; the project uses the instructor-approved yfinance fallback for fundamentals.
- Synthetic disclosure timestamps are conservative assumptions, not exact KAP timestamps.
- TCMB/TUIK/FED official live APIs are not required for the demo; curated point-in-time macro records are used with yfinance FX.
- RSS news context is metadata/evidence only and is not a standalone trade signal.
- Instructor-approved public video/STT is not implemented.
- Strategy variants A-C are measured; D-E are unavailable until executable macro/news-video signals are defined.
- Harness A-E comparison is deterministic capability scoring; live Gemini explanation exists separately.
- This is not investment advice and has no broker integration.
